```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns

sns.set_theme(style="whitegrid")

df = pd.read_csv("../data/E-commerce_Customer_Segmentation_2026.csv")
print(f"Dataset Shape: {df.shape[0]} rows, {df.shape[1]} columns\n")
print("Data Types & Missing Values:")
df.info()
df.head()
```

```python
df=df.drop(columns=['customer_id', 'rfm_score', 'customer_health_score','churn_risk_score','rfm_category','health_status','satisfaction_level'])
df.shape
```

```python
df.isnull().sum()
```

```python
df['preferred_category_2'] = df['preferred_category_2'].fillna('Unknown')
df['preferred_category_3'] = df['preferred_category_3'].fillna('Unknown')
df.isnull().sum()
```

```python
df[['preferred_category_2', 'preferred_category_3']].isnull().sum()
    
```

```python
df.describe()
```

```python
df = pd.get_dummies(df, columns=['preferred_payment_method'],drop_first=True)
```

```python
# Check unique value counts for all non-numeric columns
df.select_dtypes(include=['object', 'category']).nunique()
```

```python
df = df.drop(columns=['social_media_presence', 'country', 'city'])
```

```python
categorial_columns = df.select_dtypes(include=['object', 'category']).columns

df = pd.get_dummies(df, columns=categorial_columns,drop_first=True)

df.shape
df.columns
```

```python
x = df.drop(columns=['satisfaction_score'])
y = df['satisfaction_score']
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(X_train_scaled, y_train)
```

```python
y_pred = model.predict(X_test_scaled)
y_pred = model.predict(X_test_scaled)
```

```python
from sklearn.metrics import mean_squared_error

mse = mean_squared_error(y_test, y_pred)
print("MSE:", mse)
```

```python
import pandas as pd

# Match feature names with their learned coefficients
feature_importances = pd.Series(model.coef_, index=x.columns)

# Display the top 5 largest positive coefficients
print(feature_importances.abs().sort_values(ascending=False).head(5))
```

```python
from sklearn.metrics import r2_score

r2 = r2_score(y_test, y_pred)
print("R2 Score:", r2)
```

```python
from sklearn.ensemble import RandomForestRegressor

rf_model = RandomForestRegressor(n_estimators=100, random_state=42)

rf_model.fit(X_train_scaled, y_train)

y_pred_rf = rf_model.predict(X_test_scaled)

rf_mse = mean_squared_error(y_test, y_pred_rf)
rf_r2 = r2_score(y_test, y_pred_rf)

print("Random Forest MSE:", rf_mse)
print("Random Forest R2 Score:", rf_r2)


```

```python
rf_importances = pd.Series(rf_model.feature_importances_, index=x.columns)
print(rf_importances.abs().sort_values(ascending=False).head(5))
```
